# Experiment 1: TF-IDF
***

**T**erm **F**requency - **I**nverse **D**ocument **F**requency (TF-IDF) is a traditional NLP technique to look at words that appear in both pieces of text, and score them based on how often they appear. For this experiment, you are free to use the TF-IDF implementation provided by scikit-learn.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

nltk.download('punkt')
stop_words = set(stopwords.words('english'))
```

```python
query = 'fruits'
# query  terms: 'fruits' / 'vegetables' / 'healthy foods in Canada'

# vectorizer = TfidfVectorizer(stop_words=stop_words)
# Cast the set/dict to a list
vectorizer = TfidfVectorizer(stop_words=list(stop_words))

vectors = vectorizer.fit_transform([query] + documents)
print(vectors)

# Calculate the word frequency, and a measure of similarity of the search terms with each document.
# Do not apply any text pre-processing (i.e. cleanup) yet.
```

```python
#FRUITS
#similarity score
from sklearn.metrics.pairwise import linear_kernel
similarity = linear_kernel(vectors[0:1], vectors[1:]).flatten()

import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

#count each word
count_vectorizer = CountVectorizer(stop_words=list(stop_words))
X = count_vectorizer.fit_transform(documents)

feature_names = count_vectorizer.get_feature_names_out()
query_terms = query.lower().split()

word_frequencies = []

#should show title, score, and freq of a word in q
for i, title in enumerate(titles):
    freq_info = {"Title": title}
    for term in query_terms:
        if term in feature_names:
            term_index = list(feature_names).index(term)
            freq_info[f"freq_{term}"] = X[i, term_index]
        else:
            freq_info[f"freq_{term}"] = 0
    word_frequencies.append(freq_info)

freq_df = pd.DataFrame(word_frequencies)

results_df = pd.DataFrame({
    "Title": titles,
    "Similarity": similarity
})

results_df = results_df.merge(freq_df, on="Title")

top5 = results_df.sort_values(by="Similarity", ascending=False).head(5)

print("Top 5 most similar documents for fruits:")
print(top5.to_string(index=False))

# results_fruits = []
# for sentence in documents:
#     sentence_results = []
#     for s in sentence:
#         sentence_results.append(nltk.word_tokenize(sentence))
#     results_fruits.append(sentence_results)
# vect = CountVectorizer(tokenizer=results_fruits)

# for each query, output the similarity scores for the top 5 documents with
# the highest score, and interpret your results
######################################################################################
##  Based on the results below, we can see that the most simiar documents to the query "fruits" were Food classes, Canada's Food Guide, Pomegranate Arakta, Pomegranate Bhagwa
##  and Contact Us. There seems to be fewer exact matches and a frequency of one occurrence of the word "fruits" in the top two documents. The documents that do indicate frequencies
##  make sense in the context of their titles. They would most likely contain exact matches of "fruits".
```

```python
# e.g. you can use a lemmatizer to reduce words down to their
# simplest 'lemma' (helpful when dealing with plurals)

from nltk import word_tokenize
from nltk.stem import WordNetLemmatizer
import re

nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
def standardize_text(text):
    text = re.sub(r'[^a-z0-9\s]', '', text.lower())
    tokens = text.split()
    #remove stop words
    lemmatized = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return ' '.join(lemmatized)
```

```python
import logging
import json
import logging
from re import sub
from multiprocessing import cpu_count

import numpy as np

import gensim.downloader as api
from gensim.utils import simple_preprocess
from gensim.corpora import Dictionary
from gensim.models import TfidfModel
from gensim.similarities import WordEmbeddingSimilarityIndex
from gensim.similarities import SparseTermSimilarityMatrix
from gensim.similarities import SoftCosineSimilarity
```

```python
# optional, but it helps
import logging

# Initialize logging.
logging.basicConfig(format='%(asctime)s : %(levelname)s : %(message)s', level=logging.WARNING)
```

```python
import nltk

# Import and download stopwords from NLTK.
nltk.download('stopwords')  # Download stopwords list.
stopwords = set(nltk.corpus.stopwords.words("english"))
```

```python
def preprocess(doc):
    # Tokenize, clean up input document string
    doc = sub(r'<img[^<>]+(>|$)', " image_token ", doc)
    # you may decide to add additional steps here
    return [token for token in simple_preprocess(doc, min_len=0, max_len=float("inf")) if token not in stopwords]
```

```python
# Load test data
with open('sample_repository.json') as in_file:
    repo_data = json.load(in_file)

titles = [item[0] for item in repo_data['data']]
documents = [item[1] for item in repo_data['data']]
```

```python
query_s = "fruits"

# Preprocess the documents, including the query string
corpus = [preprocess(document) for document in documents]
query = preprocess(query_s)
```

```python
# Download and load the GloVe word vector embeddings
if 'glove' not in locals():  # only load if not already in memory
    glove = api.load("glove-wiki-gigaword-50")

similarity_index = WordEmbeddingSimilarityIndex(glove)
```

```python
# Build the term dictionary, TF-idf model
# Keep in mind that the search query must be in the dictionary as well, in case the terms do not overlap with the documents
dictionary = Dictionary(corpus+[query])
tfidf = TfidfModel(dictionary=dictionary)

# Create the term similarity matrix.
# The nonzero_limit enforces sparsity by limiting the number of non-zero terms in each column.
# In my case, I got best results by removing the default value of 100
similarity_matrix = SparseTermSimilarityMatrix(similarity_index, dictionary, tfidf)  # , nonzero_limit=None)
```

```python
# Compute similarity measure between the query and the documents.
query_tf = tfidf[dictionary.doc2bow(query)]

index = SoftCosineSimilarity(
            tfidf[[dictionary.doc2bow(document) for document in corpus]],
            similarity_matrix)

doc_similarity_scores = index[query_tf]
```

```python
# for each query, output the similarity scores for the top 5 documents with
# the highest score, and interpret your results
fruits_results = sorted(zip(titles, doc_similarity_scores), key=lambda x: x[1], reverse=True)

for rank, (title, score) in enumerate(fruits_results[:5], start=1):
        print(f"{rank}. {title} - Similarity Score: {score:.4f}")
```

```python
#interpret your results

##  Based on the ranking and similarity scores of the top documents, the matching was on the higher end (close to 0.8 or higher) indicating higher similarty based on semantics. The relation
##  between words plays a role here, so the matching would naturally be higher as there would be more words in general that would "relate" to a query rather than exactly "match" it.
```
